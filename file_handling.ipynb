```python
def count_words():
    with open("document.txt", "r") as file:
        content = file.read()
        words = content.split()
        return len(words)

result = count_words()
print("Total number of words:", result)
```

```python
def find_replace(old_word, new_word):
    with open("data.txt", "r") as file:
        content = file.read()

    content = content.replace(old_word, new_word)

    with open("data.txt", "w") as file:
        file.write(content)

    print("Word replaced successfully!")

find_replace("Python", "Java")
```

```python
def reverse_lines():
    with open("reverse.txt", "r") as file:
        for line in file:
            print(line.rstrip()[::-1])

reverse_lines()
```

```python
def count_stats():
    with open("stats.txt", "r") as file:
        content = file.read()

    lines = len(content.splitlines())
    words = len(content.split())
    characters = len(content)

    return lines, words, characters

l, w, c = count_stats()

print("Lines:", l)
print("Words:", w)
print("Characters:", c)
```
